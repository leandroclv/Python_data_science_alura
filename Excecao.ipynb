###Exceções

- Erros detectados durante a execução do programa
- Interrompem o fluxo do programa quando não tratadas


####Tipos de Exceções

- SyntaxError -> Ocorre quando é detectado pelo parser(analisador) um erro na descrição do código
- NameError -> Exceção lançada quando tentamos utilizar um nome de algum elemento que não está presente em nosso código
- IndexError -> Exceção lançada quando tentamos indexar alguma estrutura de dados como lista, tupla ou até string além de seus limites
- TypeError -> Exceção lançada quando um operador ou função são aplicados a um objeto cujo tipo é inapropriado
- KeyError -> Exceção lançada quando tentamos acessar uma chave que não está no dicionário presente em nosso código.
- Warning -> Exceção lançado em situações que precisamos alertar à pessoa usuária sobre algumas condições do código. Essas condições não necessariamente interrompem a execução do programa, mas podem lançar avisos sobre uso de módulos obsoletos, ou que possam ser depreciados em atualizações futuras ou também para alterações que podem reverberar sobre alguma parte do código.

```python
notas = {'João': [8.0, 9.0, 10.0], 'Maria': [9.0, 7.0, 6.0], 'José': [3.4, 7.0, 8.0],'Cláudia': [5.5, 6.6, 8.0], 'Ana': [6.0, 10.0, 9.5], 'Joaquim': [5.5, 7.5, 9.0], 'Júlia': [6.0, 8.0, 7.0], 'Pedro': [3.0, 4.0, 6.0]}
```

```python
nome = input('Digite o nome do estudante: ')
resultado = notas[nome]
resultado
```

```python
try:
    nome = input('Digite o nome do estudante: ').capitalize()
    resultado = notas[nome]
    print(resultado)
except Exception as e:
    print('Estudante não matriculado na turma')
```

```python
#try:
    #Código a ser executado. Caso uma exceção seja lançada, para imediatamente
#except:
    #Se uma exceção for lançada no try, rode esse código, senão pule esta etapa
#else:
    #Se não houver uma exeção lançada pelo try, rode essa parte
```

```python
try:
    nome = input('Digite o nome do estudante: ').capitalize()
    resultado = notas[nome]
    #print(resultado)
except Exception as e:
    print('Estudante não matriculado na turma')
else:
    print(resultado)
```

```python
#Finally
#try:
    #Código a ser executado. Caso uma exceção seja lançada, para imediatamente
#except:
    #Se uma exceção for lançada no try, rode esse código, senão pule esta etapa
#else:
    #Se não houver uma exeção lançada pelo try, rode essa parte
#finally:
    #Rode essa parte (com ou sem exceção)
```

```python
try:
    nome = input('Digite o nome do estudante: ').capitalize()
    resultado = notas[nome]
    #print(resultado)
except Exception as e:
    print('Estudante não matriculado na turma')
else:
    print(resultado)
finally:
    print('A consulta foi encerrada!')
```
