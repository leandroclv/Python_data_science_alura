###Tratando dados nulos

```python
utf = 'https://raw.githubusercontent.com/alura-cursos/pandas-conhecendo-a-biblioteca/main/base-de-dados/aluguel.csv'

df = pd.read_csv(utf, sep=';')
df
```

```python
df.isnull() #True -> dados nulos e False -> dados não nulos
```

```python
df.isnull().sum() #Soma a quantidade de dados nulos existentes True = 1 e False = 0
```

```python
df.fillna(0)
```
